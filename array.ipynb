```python
nilai=[80,90,100,70,89]

nama=['Dedi','Husna','Jaka','Ilham','Arya']

data=['dedi', 90, 'Cikarang','Bekasi',76]
```

```python
#Husna mempunyai nilai 90, yang tinggal di Cikarang

print(f"{nama[1]} mempunyai nilai {nilai[1]}, yang tinggal di {data[2]}")
```

```python
#Akses nilai dengan index negatif
#Nama saya Ilham tinggal di Bekasi
print(f"Nama saya {nama[-2]} tinggal di {data[-2]}")
```

```python
data[3]="Bantar Gebang"
```

```python
#Karena Bunga telat dapat nilai 60, tinggal di Graha Asri
nama.append("Bunga")
nilai.append(60)
data.append("Graha Asri")
print(f"Karena {nama[-1]} telat dapat nilai {nilai[-1]},tinggal di {data[-1]}")
```

```python
#Nama Rian  dapat nilai 100, tinggal di Rawa Banteng
nama.insert(-2,"Rian")
nilai.insert(-2,100)
data.insert(-2,"Rawa Banteng")
print(f"Nama {nama[-2]}  dapat nilai {nilai[-2]},tinggal di {data[-2]}")
```
